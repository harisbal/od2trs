```python
cols = ['OrigZoneId', 'DestZoneId', 'Purpose', 'TimePeriod']
df1['tmp'] = df1[cols].values.sum(axis=1)
df2['tmp'] = df2[cols].values.sum(axis=1)
```

```python
dfg1 = df1.groupby(level=0)['tmp'].sum()
dfg2 = df2.groupby(level=0)['tmp'].sum()
```

```python
t1 = dfg1.reset_index()
t2 = dfg2.reset_index()

t1.set_index('tmp', inplace=True)
t2.set_index('tmp', inplace=True)
```

```python
jn = t1.join(t2, lsuffix='_orig', rsuffix='_reverse')
```

```python
l = pd.merge(t1, t2, how='outer', left_index=True, right_index=True)
l
```

```python
d = dict(zip(jn.TripChainId_orig, jn.TripChainId_reverse))
```

```python
d
```

```python
tc_map = pd.DataFrame.from_dict(d, orient='index', )
tc_map.index.name = 'tc_id_original'
tc_map.columns = ['tc_id_reverse']
tc_map = tc_map.squeeze()
tc_map
```

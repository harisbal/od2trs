```python
%store -r tcs_defs_optim
%store -r tcs_defs_orig
```

```python
tcs_defs_optim = tcs_defs_optim.copy()
tcs_defs_orig = tcs_defs_orig.copy()
```

```python
tcs_defs_orig.head(2)
```

```python
tcs_defs_optim.head(2)
```

```python
tcs_defs_orig['LegDescr'] = tcs_defs_orig.apply(lambda x: '_'.join(x), axis=1)
tcs_defs_orig = tcs_defs_orig[['LegDescr']]
tcs_defs_orig.head(2)
```

```python
tcs_defs_optim['LegDescr'] = tcs_defs_optim.apply(lambda x: '_'.join(x), axis=1)
tcs_defs_optim = tcs_defs_optim[['LegDescr']]
tcs_defs_optim.head(2)
```

```python
dfg_orig = tcs_defs_orig.groupby('TripChainId').sum().reset_index()
dfg_optim = tcs_defs_optim.groupby('TripChainId').sum().reset_index()
```

```python
jn = pd.merge(left=dfg_orig, right=dfg_optim, on='LegDescr', how='outer', suffixes=['_Orig', '_Optim'])
jn.drop('LegDescr', axis=1, inplace=True)
```
